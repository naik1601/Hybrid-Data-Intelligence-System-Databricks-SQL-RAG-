```python
from pyspark.sql.functions import lit

branch_schema = "branch_id int, branch_country string, branch_city string"
df = spark.read.schema(branch_schema).parquet("abfss://landing@projpolicysytem.dfs.core.windows.net/BranchData/*.parquet")
display(df)


```

```python
df_with_flag = df.withColumn("merge_flag", lit(False))
bronze_path = "abfss://bronzelayer@projpolicysytem.dfs.core.windows.net/Branch"
df_with_flag.write \
    .format("delta") \
    .option("path", bronze_path) \
    .mode("append") \
    .saveAsTable("policyprojcatalog.policyprojdb.Branch")
```
